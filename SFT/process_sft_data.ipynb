```python
# 数一数每个通用问题有多少条 sft 数据

import os
import json
import pandas as pd


dir = r'trajectories/general_traj/'

file_path=r'D:\DissertationProject\DissertationProject\dataset\general_dataset\sft_candidates.csv'
df = pd.read_csv(file_path)

level_1_line = 0
level_2_line = 0
level_3_line = 0
level_1_data = 0
level_2_data = 0
level_3_data = 0
total_count = 0
for file in os.listdir(dir):
    file_name=file.split('.')[0]
    task_id=file_name.split('task')[1]
    task_id=int(task_id)
    df_row=df.iloc[task_id-1]
    
    if file.endswith('.jsonl'):
        with open(os.path.join(dir, file), 'r', encoding='utf-8') as f:
            count = 0
            for line in f:
                count += 1
            print(file, count, '-difficulty-', df_row['difficulty'])
    if df_row['difficulty'] == 1:
        level_1_line += count
        level_1_data += 1
    elif df_row['difficulty'] == 2:
        level_2_line += count
        level_2_data += 1
    else:
        level_3_line += count
        level_3_data += 1
    total_count += count
print('Total:', total_count)
print('Level 1 data:', level_1_data)
print('Level 2 data:', level_2_data)
print('Level 3 data:', level_3_data)
print('Level 1 line:', level_1_line, level_1_line/total_count)
print('Level 2 line:', level_2_line, level_2_line/total_count)
print('Level 3 line:', level_3_line, level_3_line/total_count)


    
```

```python
# 数一数每个 CS qa 有多少条 sft 数据

import os
import json
import pandas as pd


dir = r'trajectories/CS_traj/'

level_1_line = 0
level_2_line = 0
level_3_line = 0
level_1_data = 0
level_2_data = 0
level_3_data = 0
total_count = 0
for file in os.listdir(dir):
    if file.endswith('.jsonl'):
        with open(os.path.join(dir, file), 'r', encoding='utf-8') as f:
            count = 0
            for line in f:
                count += 1
            if count <= 10: difficulty = 1
            elif count <= 20: difficulty = 2
            else: difficulty = 3
            print(file, count, '-difficulty-', difficulty)
            if difficulty == 1:
                level_1_line += count
                level_1_data += 1
            elif difficulty == 2:
                level_2_line += count
                level_2_data += 1
            else:
                level_3_line += count
                level_3_data += 1
    total_count += count
print('Total:', total_count)
print('Level 1 data:', level_1_data)
print('Level 2 data:', level_2_data)
print('Level 3 data:', level_3_data)
print('Level 1 line:', level_1_line, level_1_line/total_count)
print('Level 2 line:', level_2_line, level_2_line/total_count)
print('Level 3 line:', level_3_line, level_3_line/total_count)


    
```

```python
# 通用 deep research sft 和 CS qa sft

import os
import json
import random
import pandas as pd

general_dir = r'trajectories/general_traj/'
CS_dir = r'trajectories/CS_traj/'

general_trajs = []
for file in os.listdir(general_dir):
    if file.endswith('.jsonl'):
        with open(os.path.join(general_dir, file), 'r', encoding='utf-8') as f:
            general_trajs_per_qa = [json.loads(line) for line in f]
    general_trajs.extend(general_trajs_per_qa)

print('General trajs:', len(general_trajs))

cs_trajs = []
for file in os.listdir(CS_dir):
    if file.endswith('.jsonl'):
        with open(os.path.join(CS_dir, file), 'r', encoding='utf-8') as f:
            cs_trajs_per_qa = [json.loads(line) for line in f]
    cs_trajs.extend(cs_trajs_per_qa)
    
print('CS trajs:', len(cs_trajs))

with open(r'data/cs_sft_data.jsonl', 'w', encoding='utf-8') as f:
    for traj in cs_trajs:
        f.write(json.dumps(traj, ensure_ascii=False) + '\n')
        
with open(r'data/general_sft_data.jsonl', 'w', encoding='utf-8') as f:
    for traj in general_trajs:
        f.write(json.dumps(traj, ensure_ascii=False) + '\n')
```

```python
# 计算r'trajectories/general_traj/'一条数据的输入token数
import os
import json
import tiktoken

# 初始化编码器 (使用 GPT-4 的编码器)
encoder = tiktoken.get_encoding("cl100k_base")

def count_input_tokens(file_path, line_index=0):
    """
    计算 JSONL 文件中指定行的输入 token 数
    
    Args:
        file_path: JSONL 文件路径
        line_index: 要读取的行号（从 0 开始）
    
    Returns:
        输入 token 数
    """
    with open(file_path, 'r', encoding='utf-8') as f:
        data = [json.loads(line) for line in f]
        data = data[line_index]
        # print(data)
        # 提取 messages 中的 user 和 assistant 内容作为输入
        messages = data.get('messages', [])
        # print(messages)
        
        # 计算所有消息的 token 数（输入包括 user 和 assistant 的消息）
        total_tokens = 0
        for msg in messages:
            content = msg.get('content', '')
            tokens = encoder.encode(content)
            total_tokens += len(tokens)
            print(f"角色：{msg.get('role', 'unknown')}, Token 数：{len(tokens)}")
        
        return total_tokens
    return 0

# 使用示例
dir_path = r'trajectories/general_traj/'
files = [f for f in os.listdir(dir_path) if f.endswith('.jsonl')]

if files:
    # 选择第一个文件的第一条数据
    selected_file = os.path.join(dir_path, files[4])
    print(f"分析文件：{selected_file}")
    print("=" * 60)
    
    token_count = count_input_tokens(selected_file, line_index=2)
    
    print("=" * 60)
    print(f"总输入 token 数：{token_count}")
else:
    print("未找到 JSONL 文件")
```

```python
# 分析 sft log 数据


import json
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

def analyze_and_plot_academic_logs(log_path, output_dir="."):
    # ---------------------------------------------------------
    # 1. 全局学术图表样式设置 (Academic rcParams)
    # ---------------------------------------------------------
    plt.rcParams.update({
        "font.family": "serif",             # 学术常用的衬线字体家族
        "font.serif": ["Times New Roman", "DejaVu Serif"], 
        "font.size": 12,
        "axes.labelsize": 14,               # 轴标签稍大
        "axes.titlesize": 14,
        "xtick.labelsize": 12,
        "ytick.labelsize": 12,
        "legend.fontsize": 11,
        "legend.frameon": True,             # 保留图例边框
        "legend.edgecolor": "black",
        "legend.fancybox": False,           # 直角边框更显学术
        "axes.linewidth": 1.2,              # 坐标轴线变粗
        "lines.linewidth": 2.0,             # 数据线宽
        "xtick.major.width": 1.2,
        "ytick.major.width": 1.2,
        "xtick.direction": "in",            # 刻度线朝内
        "ytick.direction": "in",
        "figure.dpi": 300,                  # 高分辨率
        "savefig.dpi": 300,
        "savefig.bbox": "tight",            # 保存时去除边缘白边
    })

    train_data = []
    eval_data = []
    
    # ---------------------------------------------------------
    # 2. 解析 JSONL 日志文件
    # ---------------------------------------------------------
    with open(log_path, 'r', encoding='utf-8') as f:
        for line in f:
            if not line.strip():
                continue
            try:
                # 兼容带有 格式的日志行
                clean_line = line.split("] ", 1)[-1] if "] {" in line else line
                data = json.loads(clean_line)
                
                if "global_step/max_steps" in data:
                    step = int(data["global_step/max_steps"].split("/")[0])
                    
                    if "loss" in data:
                        train_data.append({
                            "step": step,
                            "train_loss": data["loss"],
                            "train_token_acc": data["token_acc"]
                        })
                    elif "eval_loss" in data:
                        eval_data.append({
                            "step": step,
                            "eval_loss": data["eval_loss"],
                            "eval_token_acc": data["eval_token_acc"]
                        })
            except json.JSONDecodeError:
                continue

    df_train = pd.DataFrame(train_data)
    df_eval = pd.DataFrame(eval_data)

    # ---------------------------------------------------------
    # 3. 绘制学术级别双子图 (1x2 水平排版适合双栏或跨栏论文)
    # ---------------------------------------------------------
    # 图表尺寸设置：宽 10 英寸，高 4 英寸（适合论文跨栏展示）
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
    
    # 学术配色方案 (深蓝与深红，高对比度)
    COLOR_TRAIN = "#1f77b4" # 沉稳的蓝色
    COLOR_EVAL = "#d62728"  # 砖红色

    # -------- 子图 1: Loss 曲线 --------
    # 训练集数据点多，用较浅颜色的实线，不加 Marker
    ax1.plot(df_train['step'], df_train['train_loss'], 
             label='Train Loss', color=COLOR_TRAIN, alpha=0.7, linestyle='-')
    
    # 验证集数据点少且关键，用显眼的颜色、虚线和形状标记 (Marker)
    ax1.plot(df_eval['step'], df_eval['eval_loss'], 
             label='Eval Loss', color=COLOR_EVAL, linestyle='--', 
             marker='o', markersize=6, markerfacecolor='white', markeredgewidth=1.5)
    
    ax1.set_xlabel('Global Steps')
    ax1.set_ylabel('Loss')
    ax1.grid(True, linestyle=':', alpha=0.6) # 使用点状弱网格线
    ax1.legend(loc='upper right')

    # -------- 子图 2: Accuracy 曲线 --------
    ax2.plot(df_train['step'], df_train['train_token_acc'], 
             label='Train Accuracy', color=COLOR_TRAIN, alpha=0.7, linestyle='-')
    
    ax2.plot(df_eval['step'], df_eval['eval_token_acc'], 
             label='Eval Accuracy', color=COLOR_EVAL, linestyle='--', 
             marker='s', markersize=6, markerfacecolor='white', markeredgewidth=1.5) # 使用正方形 Marker (s)
    
    ax2.set_xlabel('Global Steps')
    ax2.set_ylabel('Token Accuracy')
    
    # 将 y 轴格式化为百分比，更符合学术规范
    ax2.yaxis.set_major_formatter(ticker.PercentFormatter(xmax=1.0, decimals=0))
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='lower right')

    # ---------------------------------------------------------
    # 4. 优化布局并导出 (同时导出 PDF 和 高清 PNG)
    # ---------------------------------------------------------
    plt.tight_layout()
    
    pdf_path = os.path.join(output_dir, "academic_training_metrics.pdf")
    png_path = os.path.join(output_dir, "academic_training_metrics.png")
    
    # PDF 是论文排版的首选（无限放大不失真）
    plt.savefig(pdf_path, format='pdf')
    # PNG 备用（用于 PPT 汇报或快速预览）
    plt.savefig(png_path, format='png', dpi=300)
    
    print(f"✅ 学术图表已成功生成！")
    print(f"📄 矢量图 (LaTeX首选): {pdf_path}")
    print(f"🖼️ 高清位图 (PPT/预览): {png_path}")

    plt.close()

# 执行脚本（确保同级目录下有 logging.jsonl 文件）
if __name__ == "__main__":
    analyze_and_plot_academic_logs("sft_log/logging.jsonl")
```
